  3. Berapa total pendapatan dari seluruh menu yang terjual?
* **Dugaan Masalah Kualitas Data:**
  * Missing value pada kolom `menu` dan `terjual`

* **Rencana Teknik Pembersihan:**
  * Menghapus baris yang memiliki nilai kosong di kolom `menu` (`dropna`)

```python
import pandas as pd

# ==========================================
# 1. DATA LOADING & INSPECTION
# ==========================================
df = pd.read_csv('data_kantin.csv')

print("--- 5 Baris Pertama Data ---")
print(df.head())

print("\n--- Informasi Struktur Dataset ---")
print(df.info())

print("\n--- Statistik Ringkas ---")
print(df.describe())

print("\n--- Ukuran Dataset (Baris, Kolom) ---")
print(df.shape)


# ==========================================
# 2. DATA CLEANING
# ==========================================
# Cek dan hapus duplikat
print("\nJumlah duplikat:", df.duplicated().sum())
df = df.drop_duplicates()

# Cek missing value
print("\nMissing value per kolom:\n", df.isnull().sum())

# Tangani missing value
df = df.dropna(subset=['menu'])
df['terjual'] = df['terjual'].fillna(0)

# Penyesuaian tipe data
df['terjual'] = df['terjual'].astype(int)
df['tanggal'] = pd.to_datetime(df['tanggal'])


# ==========================================
# 3. DATA MANIPULATION
# ==========================================
# a. Filtering: Mengambil transaksi dengan penjualan di atas 20 porsi
df_laris = df[df['terjual'] > 20]

# b. Sorting: Mengurutkan transaksi dari jumlah terjual terbanyak
df_sorted = df.sort_values(by='terjual', ascending=False)

# c. Kolom Turunan: Menghitung total pendapatan per transaksi
df['total_pendapatan'] = df['harga'] * df['terjual']

# d. Groupby / Agregasi: Total pendapatan per menu
df_grouped = df.groupby('menu')['total_pendapatan'].sum().reset_index().sort_values(by='total_pendapatan', ascending=False)
print("\n--- Total Pendapatan per Menu ---")
print(df_grouped)


# ==========================================
# 4. SIMPAN DATASET BERSIH
# ==========================================
df.to_csv('dataset_bersih.csv', index=False)
print("\nDataset bersih berhasil disimpan sebagai 'dataset_bersih.csv'")
```
